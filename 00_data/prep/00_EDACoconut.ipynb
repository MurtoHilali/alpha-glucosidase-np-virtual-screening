```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('coconut_cleaned.csv')
data.head()
```

Here, let's get a view of the distribution of some of the numerical columns in COCONUT, including the distribution of molecular weight, heavy atom count, and drug-likeness.

```python
numerical_columns = [
    'total_atom_count', 
    'heavy_atom_count', 
    'molecular_weight', 
    'alogp', 
    'topological_polar_surface_area', 
    'rotatable_bond_count', 
    'hydrogen_bond_acceptors',
    'hydrogen_bond_donors',
    'lipinski_rule_of_five_violations',
    'aromatic_rings_count',
    'qed_drug_likeliness',
    'np_likeness'
]

# display column histograms in a grid
num_cols = 3
num_rows = (len(numerical_columns) + num_cols - 1) // num_cols
fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, num_rows * 4))
axes = axes.flatten()

for i, key in enumerate(numerical_columns):
    axes[i].hist(data[key], color='blue', alpha=0.9)
    axes[i].axvline(data[key].mean(), color='red', linestyle='dashed', linewidth=1)
    axes[i].set_title(f'{key}')
    axes[i].set_xlabel(key)
    axes[i].set_ylabel('Frequency')

fig.subplots_adjust(hspace=0.5, wspace=0.5)
```

```python
# Let's take a look at the categorical columns as well, specifically the chemical classes.

## We'll start plotting the value_counts for the 'chemical_super_class' column.

data['chemical_super_class'].value_counts().plot(kind='bar', figsize=(10,6), color='green', alpha=0.7)
```

```python
data['hba_to_hbd_ratio'] = data['hydrogen_bond_acceptors'] / (data['hydrogen_bond_donors'] + 1e-6) # avoid division by zero
```

```python
plt.hist(data['hba_to_hbd_ratio'], bins=50, color='purple', alpha=0.8)
plt.axvline(data['hba_to_hbd_ratio'].mean(), color='red', linestyle='dashed', linewidth=1)
plt.title('Hydrogen Bond Acceptor to Donor Ratio Distribution')
plt.xlabel('H-Bond Acceptor to Donor Ratio')
plt.ylabel('Frequency')
plt.show()
```

```python
data['polar_surface_density'] = data['topological_polar_surface_area'] / data['molecular_weight']
plt.hist(data['polar_surface_density'], bins=50, color='orange', alpha=0.8)
plt.axvline(data['polar_surface_density'].mean(), color='red', linestyle='dashed', linewidth=1)
plt.title('Polar Surface Area to Molecular Weight Ratio Distribution')
plt.xlabel('Polar Surface Area / Molecular Weight')
plt.ylabel('Frequency')
plt.show()
```

```python
data['polar_flexibility_burden'] = data['topological_polar_surface_area'] / (data['rotatable_bond_count'] + 1e-6)  # avoid division by zero
plt.hist(data['polar_flexibility_burden'], bins=50, color='cyan', alpha=0.8)
plt.axvline(data['polar_flexibility_burden'].mean(), color='red', linestyle='dashed', linewidth=1)
plt.title('Polar Flexibility Burden Distribution')
plt.xlabel('Polar Flexibility Burden')
plt.ylabel('Frequency')
plt.show()
```

```python
data['normalized_flexibility'] = data['rotatable_bond_count'] / data['heavy_atom_count'] # aka f_rot, or fraction of rotatable bonds
plt.hist(data['normalized_flexibility'], bins=50, color='magenta', alpha=0.8)
plt.axvline(data['normalized_flexibility'].mean(), color='red', linestyle='dashed', linewidth=1)
plt.title('Normalized Flexibility Distribution')
plt.xlabel('Normalized Flexibility (Rotatable Bonds / Heavy Atom Count)')
plt.ylabel('Frequency')
plt.show()
```

```python
data['3d_vs_flatness'] = data['fractioncsp3'] / (data['aromatic_rings_count'] + 1e-6)  # avoid division by zero
plt.hist(data['3d_vs_flatness'], bins=50, color='brown', alpha=0.8)
plt.axvline(data['3d_vs_flatness'].mean(), color='red', linestyle='dashed', linewidth=1)
plt.title('3D vs Flatness Distribution')
plt.xlabel('3D vs Flatness Ratio')
plt.ylabel('Frequency')
plt.show()
```

```python
data['packing_efficiency'] = data['van_der_walls_volume'] / data['molecular_weight']
plt.hist(data['packing_efficiency'], bins=50, color='teal', alpha=0.8)
plt.axvline(data['packing_efficiency'].mean(), color='red', linestyle='dashed', linewidth=1)
plt.title('Packing Efficiency Distribution')
plt.xlabel('Van der Waals Volume / Molecular Weight')
plt.ylabel('Frequency')
plt.show()
```
